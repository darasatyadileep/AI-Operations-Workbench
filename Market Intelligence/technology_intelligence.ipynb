```python
TECHNOLOGY_THEMES = {
    "AI_ML": [
        "artificial intelligence",
        "machine learning",
        "predictive analytics",
        "AI-powered"
    ],

    "GENERATIVE_AI": [
        "generative AI",
        "GenAI",
        "large language model",
        "LLM",
        "copilot"
    ],

    "AGENTIC_AI": [
        "agentic AI",
        "AI agent",
        "autonomous agent"
    ],

    "AUTONOMOUS_OPERATIONS": [
        "autonomous operations",
        "autonomous control",
        "closed-loop optimization",
        "remote operations"
    ],

    "ROBOTICS": [
        "robot",
        "robotics",
        "robotic inspection"
    ],

    "DRONES": [
        "drone",
        "UAV",
        "unmanned aerial"
    ],

    "COMPUTER_VISION": [
        "computer vision",
        "image recognition",
        "visual inspection"
    ],

    "DIGITAL_TWIN": [
        "digital twin",
        "digital twins"
    ],

    "DATA_CLOUD": [
        "cloud",
        "data platform",
        "industrial data",
        "edge computing"
    ],

    "TECHNOLOGY_STRATEGY": [
        "digital transformation",
        "technology strategy",
        "strategic technology partnership",
        "AI strategy"
    ]
}
```

```python
def detect_themes(text):

    text_lower = text.lower()

    matched_themes = []

    for theme, keywords in TECHNOLOGY_THEMES.items():

        for keyword in keywords:

            if keyword.lower() in text_lower:
                matched_themes.append(theme)
                break

    return matched_themes
```

```python
sample = """
SLB and Shell announced an agentic AI collaboration
to integrate subsurface, drilling and production workflows.
"""

detect_themes(sample)
```
